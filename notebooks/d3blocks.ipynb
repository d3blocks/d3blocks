# Welkom to the notebook of **D3Blocks**!


D3Blocks is the Python Library to Create Interactive and Standalone D3js Charts. Create interactive, stand-alone, and visually attractive charts that are built on the graphics of d3 javascript (d3js) but configurable with Python.

Visualizing can be the key to success in projects because it can reveal hidden insights in the data, and improve understanding. The best way to understand and explain the data is by making it interactive. Despite many visualization packages being available in Python, it remains challenging to create beautiful, stand-alone, and interactive charts that can also work outside your own machine. In the [d3graph blog](erdogant.medium.com), I readily showed how to create a stand-alone, interactive chart for network graphs. With D3Blocks we go beyond a single chart and created a framework that now contains 10 beautiful charts. It is open-source, and there is no need to install anything else than Python to create D3 charts. The output is an interactive chart for which you only need an (internet) browser; sharing and publishing becomes thus super easy. In the blog I introduce D3Blocks, and briefly describe the 10 blocks with hands-on examples.

In this notebook, we demonstrate that the usage of some D3Blocks. Note that not all blocks work correctly in colab notebook. This is ongoing work.

```python
from d3blocks import D3Blocks
import numpy as np

# Initialize
d3 = D3Blocks()

# Load example data
df = d3.import_example('cancer')

size = df['survival_months'].fillna(1).values / 10
opacity = np.random.randint(0, 8, df.shape[0])/10
tooltip=df['labx'].values + ' <br /> Survival: ' + df['survival_months'].astype(str).str[0:4].values
color=df['labx'].values

# Set all propreties
d3.scatter(df['tsneX'].values,                   # PC1 x-coordinates
           df['tsneY'].values,                   # PC2 y-coordinates
           tooltip=tooltip,                    # Tooltip
           color=color,   # Hex-colors or classlabels
           size=size,                          # Node size
           opacity=opacity,                    # Opacity
           stroke='#000000',
           cmap='tab20',                       # Colormap
           scale=True,                         # Scale the datapoints
           figsize=[1024, 768],
           filepath=None,
           notebook=True
           )

```

```python
from d3blocks import D3Blocks
import numpy as np

# Initialize
d3 = D3Blocks()

# Load example data
df = d3.import_example('mnist')

size = np.random.randint(0, 8, df.shape[0])
opacity = np.random.randint(0, 8, df.shape[0])/10
tooltip = df['y'].values.astype(str)

# Set all propreties
d3.scatter(df['PC1'].values,                   # PC1 x-coordinates
           df['PC2'].values,                   # PC2 y-coordinates
           x1=df['tsne_1'].values,             # tSNE x-coordinates
           y1=df['tsne_2'].values,             # tSNE y-coordinates
           color=df['tsne_2'].values.astype(str),   # Hex-colors or classlabels
           tooltip=tooltip,                    # Tooltip
           size=size,                          # Node size
           opacity=opacity,                    # Opacity
           stroke='#000000',
           cmap='tab20',                       # Colormap
           scale=True,                         # Scale the datapoints
           label_radio=['PCA', 'tSNE'],
           figsize=[1024, 768],
           filepath=None,
           notebook=True
           )

#      label         x         y        x1  ...  size   stroke  opacity tooltip
# 0        0  0.472107  0.871347  0.294228  ...     0  #000000      0.1       0
# 1        1  0.624696  0.116735  0.497958  ...     0  #000000      0.5       1
# 2        2  0.608419  0.305549  0.428529  ...     4  #000000      0.6       2
# 3        3  0.226929  0.532931  0.555316  ...     4  #000000      0.0       3
# 4        4  0.866292  0.553489  0.589746  ...     1  #000000      0.6       4
#    ...       ...       ...       ...  ...   ...      ...      ...     ...
# 1792     9  0.262069  0.709428  0.693593  ...     5  #000000      0.5       9
# 1793     0  0.595571  0.837987  0.352114  ...     6  #000000      0.5       0
# 1794     8  0.668742  0.359209  0.520301  ...     6  #000000      0.4       8
# 1795     9  0.416983  0.694063  0.683949  ...     6  #000000      0.4       9
# 1796     8  0.489814  0.588109  0.529971  ...     1  #000000      0.4       8

# [1797 rows x 12 columns]
```
